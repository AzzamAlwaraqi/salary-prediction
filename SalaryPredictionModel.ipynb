```python
#Step 5: Cleaning the Data:
#Cleaning the job title column:
#First find the uniqu and value counts 
print("jop title unique:",SalaryData['Job title'].unique())
print("jop title len:",len(SalaryData['Job title'].unique()))
SalaryData['Job title'] = SalaryData['Job title'].str.strip().str.lower()
print("jop title value count:",SalaryData['Job title'].value_counts())

#The Job title function has over 14000 Unique Value we used three Functions as filter to these Values
#First Function Catagorize the name by some Keywords

import re

#Short keywords like 'it' or 'ai' must match a whole word, otherwise 'writer' counts as IT and 'retail' as AI
def has_word(text, word):
    return re.search(r'\b' + re.escape(word) + r'\b', text) is not None

def categorize_job_title(job_title):
    
    # Check for missing or NaN values
    if pd.isna(job_title):
        return 'Uncategorized'
        
    if 'software' in job_title or 'developer' in job_title or 'programmer' in job_title:
        return 'Software/Developer'
        
    elif 'data' in job_title or 'analyst' in job_title or 'artificial' in job_title or has_word(job_title, 'ai') or 'machine learning' in job_title:
        return 'Data Analyst/Scientist/AI'
        
    elif 'project manager' in job_title:
        return 'Project Manager'

    elif 'manager' in job_title or 'director' in job_title or has_word(job_title, 'vp'):
        return 'Manager/Director/VP'
        
    elif 'sales' in job_title or 'representative' in job_title:
        return 'Sales'
        
    elif 'social media' in job_title or 'content' in job_title:
        return 'Social Media/content creator'
        
    elif 'product' in job_title or 'product designer' in job_title:
        return 'Product/Designer'
        
    elif has_word(job_title, 'hr') or 'human resources' in job_title:
        return 'HR/Human Resources'
        
    elif has_word(job_title, 'it') or 'compute' in job_title or 'computer' in job_title or 'computing' in job_title or 'tech' in job_title:
        return 'IT/Tech/computing'
        
    elif 'operations' in job_title or 'supply chain' in job_title:
        return 'Operations/Supply Chain'
        
    elif 'customer service' in job_title or 'receptionist' in job_title or 'customer' in job_title:
        return 'Customer Service/Receptionist'
    elif 'assistant' in job_title or 'secretary'  in job_title:
        return 'Customer Service/Receptionist'
        
    elif 'teacher' in job_title or 'educator' in job_title or 'professor' in job_title or 'trainer' in job_title:
        return 'Education/Trainer'
        
    elif 'medical' in job_title or 'doctor' in job_title or 'nurse' in job_title or 'health' in job_title or 'caregiver' in job_title:
        return 'Medical/Healthcare'
        
    elif 'lawyer' in job_title or 'attorney' in job_title or 'legal' in job_title or 'paralegal' in job_title or has_word(job_title, 'law'):
        return 'Legal/Law'
        
    elif 'scientist' in job_title or 'researcher' in job_title or 'r&d' in job_title or 'research' in job_title:
        return 'Research/Science'
        
    elif 'writer' in job_title or 'editor' in job_title or 'journalist' in job_title:
        return 'Writing/Editing'
        
    elif 'artist' in job_title or 'illustrator' in job_title or 'design' in job_title or 'graphic' in job_title:
        return 'Artist/designer'
        
    elif 'economy' in job_title or 'financial' in job_title or 'statistic' in job_title or 'accounting' in job_title or 'finance'in job_title or 'accountant' in job_title:
        return 'economy/finance'
        
    elif 'librarian' in job_title or 'documentation' in job_title or 'document' in job_title:
        return 'librarian/documentation'
        
    elif 'marketing' in job_title or 'brand' in job_title or 'copywriter' in job_title or 'business' in job_title:
        return 'Marketing/Business'

    elif 'administrator' in job_title or ' administrative' in job_title or 'officer' in job_title:
        return 'administrator'
        
    elif 'entrepreneur' in job_title or 'founder' in job_title or 'owner' in job_title or 'self employee' in job_title:
        return 'Entrepreneur/Owner'
        
    elif 'engineer' in job_title or 'engineering' in job_title or 'mechanical' in job_title or 'civil' in job_title or 'electrical' in job_title:
        return 'Engineer'
        
    elif 'nonprofits' in job_title or 'nonprofit' in job_title or 'donation' in job_title or 'donors' in job_title:
        return 'Nonprofit Jobs'
        
    elif 'counselor' in job_title or 'consulting' in job_title or 'advisor' in job_title:
        return 'Consultant'

    elif 'housekeeper' in job_title or 'plumber' in job_title or 'cleaner' in job_title or 'janitor' in job_title:
        return 'Cleaner/House Services'
    
    elif 'chef' in job_title or 'cook' in job_title or 'baker' in job_title:
        return 'Food/Culinary'
        
    elif 'scientist' in job_title or 'chemist' in job_title or 'biologist' in job_title:
        return 'Research/Science'
        
    elif 'driver' in job_title or 'chauffeur' in job_title or 'truck' in job_title:
        return 'Driver/Transport'
        
    elif 'photographer' in job_title or 'videographer' in job_title or 'camera' in job_title  or 'video' in job_title:
        return 'Photography/Video Production'
        
    elif 'pilot' in job_title or 'airline' in job_title or 'aviation' in job_title:
        return 'Aviation/Pilot'
        
    elif 'security' in job_title or 'guard' in job_title or 'surveillance' in job_title:
        return 'Security/Surveillance'
    
    else:
        return 'Uncategorized'

#We will Group these keyword (dictionary) and use them for the next Function (We used Ai help to group them)
category_keywords = {
    'Software/Developer': ['software', 'developer', 'programmer', 'coding'],
    'Data Analyst/Scientist/AI': ['data', 'analyst', 'scientist', 'data analysis'],
    'Manager/Director/VP': ['manager', 'director', 'vp', 'lead'],
    'Sales': ['sales', 'representative', 'account executive'],
    'Social Media/content creator': ['social media', 'content', 'creator'],
    'Product/Designer': ['product', 'designer', 'ux', 'ui'],
    'HR/Human Resources': ['hr', 'human resources', 'recruiter'],
    'Project Manager': ['project manager', 'project lead'],
    'IT/Tech/computing': ['it', 'tech', 'computing'],
    'Operations/Supply Chain': ['operations', 'supply chain', 'logistics'],
    'Customer Service/Receptionist': ['customer service', 'receptionist', 'client'],
    'Medical/Healthcare': ['doctor', 'nurse', 'healthcare', 'medical'],
    'Legal/Law': ['lawyer', 'attorney', 'legal', 'paralegal'],
    'Research/Science': ['research', 'scientist', 'lab', 'analysis'],
    'Writing/Editing': ['writer', 'editor', 'journalist', 'copywriter'],
    'Artist/designer': ['artist', 'designer', 'illustrator', 'graphic'],
    'economy/finance': ['economy', 'finance', 'accountant', 'statistician'],
    'librarian/documentation': ['librarian', 'documentation', 'archive', 'library'],
    'Marketing/Business': ['marketing', 'brand', 'business', 'advertising'],
    'Engineer': ['engineer', 'mechanical', 'electrical', 'civil'],
    'Nonprofit Jobs': ['nonprofit', 'charity', 'donation', 'foundation'],
    'Consultant': ['consultant', 'advisor', 'strategist', 'counselor'],
    'Cleaner/House Services': ['cleaner', 'housekeeper', 'janitor', 'maintenance'],
    'Food/Culinary': ['chef', 'cook', 'baker', 'restaurant'],
    'Driver/Transport': ['driver', 'chauffeur', 'truck', 'transport'],
    'Security/Surveillance': ['security', 'guard', 'surveillance', 'monitor'],
    'Aviation/Pilot': ['pilot', 'aviation', 'airline', 'flight']
}

#Second Function: Using this Function to filter the the job titles using correlation between the job titles and the catageory key words 
import difflib

def get_best_category_by_close_match(job_title, category_keywords):
    # Create a list with all category keywords combined
    all_keywords = {category: ' '.join(keywords) for category, keywords in category_keywords.items()}

    best_match_category = None
    highest_score = 0

    for category, keywords in all_keywords.items():
        matches = difflib.get_close_matches(job_title, [keywords], n=1, cutoff=0.8)
        
        if matches:
            match_score = difflib.SequenceMatcher(None, job_title, matches[0]).ratio()
            if match_score > highest_score:
                highest_score = match_score
                best_match_category = category

    return best_match_category if best_match_category else 'Other'

#Using this Function to filter the the job titles using the industry column after that if left there is still unorgnized jobs use the function above
def fallback_with_industry(row):
    if row['Job title categorized'] == 'Uncategorized':
        industry = str(row['Industry']).lower() if pd.notna(row['Industry']) else ''
        
        # Try to categorize based on the industry first
        industry_category = categorize_job_title(industry)
        if industry_category != 'Uncategorized':
            return industry_category
        
        # If not, fall back to close matching the job title using difflib
        job_title = str(row['Job title']).lower() if pd.notna(row['Job title']) else ''
        return get_best_category_by_close_match(job_title, category_keywords)
    
    return row['Job title categorized']

# Apply functions to the data
SalaryData['Job title categorized'] = SalaryData['Job title'].apply(categorize_job_title)
SalaryData['Job title categorized'] = SalaryData.apply(fallback_with_industry, axis=1)

# Check Results
uncategorized_jobs = SalaryData[SalaryData['Job title categorized'] == 'Uncategorized']
print(f"Remaining Uncategorized Count: {len(uncategorized_jobs)}")
print("Categorized Jobs Value Counts:\n", SalaryData['Job title categorized'].value_counts())
print("UnCategorized Jobs Value Counts:\n", uncategorized_jobs['Job title categorized'].value_counts())

#Finally after everything we will change the go title column with the organized column and remove the column
SalaryData['Job title'] =SalaryData['Job title categorized'].values
SalaryData=SalaryData.drop(SalaryData[SalaryData['Job title']=='Other'].index,errors='ignore')
#for checking
print("\n Checking for job title catagorization ",SalaryData['Job title'])

#drop the column 
SalaryData=SalaryData.drop('Job title categorized',axis=1,errors='ignore')

#for checking after the drop
display(SalaryData)
```

```python
#Step 5: Cleaning the Data (Gender,Age,Race,Education level,Field/Working experience):
#we will start the cleaning column by column but before that
#First we need some info about the columns in the datasets

print("Dataset Shape:\n", SalaryData.shape)
print("\nColumn Names:\n", SalaryData.columns)
print("\nNull Counts:\n", SalaryData.isnull().sum())
print("\nData Types:\n", SalaryData.dtypes)

#Clean the Gender column
SalaryData['Gender'] = SalaryData['Gender'].str.strip().str.title()

# Replace common variations with standard values
SalaryData['Gender'] = SalaryData['Gender'].replace({
    'Male': 'Male', 'Man': 'Male', 'M': 'Male',
    'Female': 'Female', 'Woman': 'Female', 'F': 'Female'
})

# Replace empty strings with NaN
SalaryData['Gender'] = SalaryData['Gender'].replace('', pd.NA)

# Calculate mode, ignoring NaN
mode_gender = SalaryData['Gender'].mode(dropna=True)[0]

# Replace any invalid or NaN values with the mode
SalaryData['Gender'] = SalaryData['Gender'].apply(lambda x: x if x in ['Male', 'Female'] else mode_gender)

# Check for remaining missing values
print("\nGender Null Counts:", SalaryData['Gender'].isnull().sum())

#########################################################################################################################
#########################################################################################################################

#Clean the Age column first before working experience and field working experience because it has the no null values and it is better to be a base
print("\n",SalaryData['Age'].unique())
print("\n",SalaryData['Age'].value_counts())
display(SalaryData[SalaryData['Age']=='under 18'])
SalaryData.drop(SalaryData[SalaryData['Age']=='under 18'].index,inplace=True)
##The reason all was choosen to drop is because a lot of them are somewhat illogaicaly consistent and while some of them are true it is
#hard to decide without factual data and their number is very little combared to data we decided to remove them

#repalcing the '65 or over' so we can get mid point
SalaryData['Age'] = SalaryData['Age'].replace({'65 or over': '65-74'})

#A function for Age and working experience and field working experience because the share the same formula 

def get_midpoint(age_range):
    start, end = map(int, age_range.split('-'))
    return (start + end) / 2


#continue
try:
    SalaryData['Age'] = SalaryData['Age'].apply(get_midpoint)
except:
    print("already done (see results)")

#see results for Age
print(SalaryData['Age'])

##########################################################################
##########################################################################


#working experience and field working experience
#checking for some values
print("\n working experience: ")
print("\n",SalaryData['Working experince'].unique())
print("\n",SalaryData['Working experince'].value_counts())
print("\n Field working experience: ")
print("\n",SalaryData['Field working experience'].unique())
print("\n",SalaryData['Field working experience'].value_counts())

#striping the column from strings
SalaryData['Working experince'] = SalaryData['Working experince'].replace({'41 years or more': '41 - 50', '1 year or less':'0 - 1'})
SalaryData['Field working experience'] = SalaryData['Field working experience'].replace({'41 years or more': '41 - 50', '1 year or less':'0 - 1'})
try:
    SalaryData['Working experince'] = SalaryData['Working experince'].map(lambda x: x.rstrip('years'))
    SalaryData['Field working experience'] = SalaryData['Field working experience'].map(lambda x: x.rstrip('years'))
except:
    print("already done (see results)")

#Start the cleaning:
try:
    SalaryData['Working experince'] = SalaryData['Working experince'].apply(get_midpoint)
    SalaryData['Field working experience'] = SalaryData['Field working experience'].apply(get_midpoint)
except:
    print("already done (see results)")

#clean the wrong values:
#First the show the worng results
display(SalaryData[SalaryData['Working experince']>=SalaryData['Age']])
display(SalaryData[SalaryData['Field working experience']>=SalaryData['Age']])

#to remove the we will apply a function that lower the mid point of these two column in the midpoint range if it is more than age
#First get the mid points
working_experience_midpoints = SalaryData['Working experince'].unique()
Field_working_experience_midpoints = SalaryData['Field working experience'].unique()

def adjust_working_experience(age, working_experience, midpoints):
    # Sort midpoints to ensure they're in ascending order
    midpoints = sorted(midpoints)
    # If working experience is greater than age, find the next lower midpoint
    if working_experience > age:
        for mid in reversed(midpoints): 
            if mid <= age:
                return mid
        return 0  
    return working_experience
#applying the Function
SalaryData['Working experince'] = SalaryData.apply(
    lambda row: adjust_working_experience(
        row['Age'], row['Working experince'], working_experience_midpoints
    ),
    axis=1
)
SalaryData['Field working experience'] = SalaryData.apply(
    lambda row: adjust_working_experience(
        row['Age'], row['Field working experience'], Field_working_experience_midpoints
    ),
    axis=1
)

#check after applying the Fuction
print("\n working experience: ")
print("\n",SalaryData['Working experince'].value_counts())
print("\n Field working experience: ")
print("\n",SalaryData['Field working experience'].value_counts())

#display the results
display(SalaryData)

#Find the correlation between the two columns
print("\n Correlation between the two column is:\n",SalaryData['Working experince'].corr(SalaryData['Field working experience']))
#The correlation is highr but no very high removing one column will be decided in the before the training 

#####################################################################################################################
#####################################################################################################################

#Cleaning of college level column
#Fisrt find the unique value and the value count
print("\nEducation level: ")
print("\n",SalaryData['Education level'].unique())
print("\n",SalaryData['Education level'].value_counts())

#Find the inconsistent values
display(SalaryData[(SalaryData['Education level']=='PhD')&(SalaryData['Age'] < 27)])
display(SalaryData[(SalaryData['Education level']=="Master's degree")&(SalaryData['Age'] < 24)])

#we will drop the phd (age less than 27) because they are ver minimal in numbers and just in case scenario
SalaryData.drop(index=24669,inplace=True,errors="ignore")

#Now fill the null values we will replace the we decide on filling it with the dominant value because they are little
SalaryData['Education level']=SalaryData['Education level'].fillna(SalaryData['Education level'].mode()[0])

############################################################################################################################
############################################################################################################################

#The Race column
SalaryData['Race']=SalaryData['Race'].str.lower()
#Find the unique value and value count
print("\nRace:")
print("\n",SalaryData['Race'].unique())
print("\n",SalaryData['Race'].value_counts())

#Race will not be cleaned because it wont be included in the model because the overwhel of white race which is more than 80% of the data
#Update:it will be cleaned because after the te evalution (Random Forest Regressor), Race affect the test and training score positivly
def categorize_race(race):
    if pd.isna(race):  # Check if the value is NaN
        return "Unknown"
    race = str(race).lower()  # Convert to lowercase for consistency
    if race == "white":
        return "White"
    elif race == "asian or asian american":
        return "Asian or Asian American"
    elif race == "black or african american":
        return "Black or African American"
    elif race == "hispanic, latino, or spanish origin":
        return "Hispanic, Latino, or Spanish Origin"
    elif race == "another option not listed here or prefer not to answer":
        return "Prefer Not to Answer"
    elif "white" in race:
        return "Mixed (Includes White)"
    elif any(x in race for x in ["asian", "black", "hispanic", "middle eastern"]):
        return "Mixed/Other"
    else:
        return "Other"

# Apply the function to the 'Race' column
SalaryData['Race'] = SalaryData['Race'].apply(categorize_race)

###############################################################################################################
###############################################################################################################

#Final cleaning of The Annual Salary Column
import numpy as np

#find the null values 
display(SalaryData[SalaryData['Annual salary (USD)'].isnull()])

#before continuing Farther Just in case transfer the column to numeric 
SalaryData['Annual salary (USD)'] = pd.to_numeric(SalaryData['Annual salary (USD)'], errors='coerce')

# replace null values
# Group by 'Job title' and apply the median for each group to fill missing values
SalaryData['Annual salary (USD)'] = SalaryData.groupby('Job title')['Annual salary (USD)'].transform(lambda x: x.fillna(x.median()))

# Check if there are still any missing values, and drop them (a job title with no salary at all has no median)
print(SalaryData['Annual salary (USD)'].isnull().sum())
SalaryData = SalaryData.dropna(subset=['Annual salary (USD)'])

#A yearly salary under 1000 USD is a typo or a monthly figure (some answers were 0), so we drop them
display(SalaryData[SalaryData['Annual salary (USD)'] < 1000])
SalaryData = SalaryData[SalaryData['Annual salary (USD)'] >= 1000]

# find outliers First method:
#Q1 and Q3 method 

#Calculate Q1 (25th percentile) and Q3 (75th percentile)
Q1 = SalaryData['Annual salary (USD)'].quantile(0.25)
Q3 = SalaryData['Annual salary (USD)'].quantile(0.75)

# Calculate IQR (Interquartile Range)
IQR = Q3 - Q1

# Define the bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliersIQR = SalaryData[(SalaryData['Annual salary (USD)'] < lower_bound) | (SalaryData['Annual salary (USD)'] > upper_bound)]

#Display the outliers
print("Outliers based on IQR method:")
print(outliersIQR[['Annual salary (USD)']])

#Drop the outliers (Check the reason below in the bottom of the page)
SalaryData=SalaryData.drop(outliersIQR.index,errors='ignore')

#############

# Z-score method 
from scipy import stats
# Calculate the Z-Score
SalaryData['z_score'] = stats.zscore(SalaryData['Annual salary (USD)'])

# Define the Z-Score threshold (3)
z_threshold = 3

# Identify outliers where Z-Score is greater than 3 or less than -3
outliersZScore = SalaryData[SalaryData['z_score'].abs() > z_threshold]

# Display the outliers
print("Outliers based on Z-Score method:")
print(outliersZScore[['Annual salary (USD)']])
#Drop the outliers (Check the reason below in the bottom of the page)
SalaryData=SalaryData.drop(outliersZScore.index,errors='ignore')

#We decided to drop outliers as a extreme metod to enhance the machine learning model score while it worked it is not certain that it is the right method
```

```python
#Step 5:Cleaning the country column:

import country_converter as coco  
from difflib import get_close_matches

#Checking for null values
display(SalaryData[SalaryData['Country'].isnull()])

#Lowercase the the column
SalaryData['Country']=SalaryData['Country'].str.lower()
#removing special character and space
SalaryData['Country'] = SalaryData['Country'].str.replace(r'[^\w\s]', '', regex=True)
#Find all the unique Values
SalaryData['Country'].unique()

valid_countries = coco.CountryConverter().data['name_short'].tolist()

# Adding two Function to clean and correct country names , we use these Functions as a filter
#Function 1:
def categorize_country(country_name):
    country_name = str(country_name).lower().strip()  
    
    # Match specific substrings or phrases for countries
    if 'usa' in country_name or 'america' in country_name or 'united states' in country_name or 'us' in country_name:
        return 'United States'
    elif 'india' in country_name:
        return 'India'
    elif 'uk' in country_name or 'united kingdom' in country_name or 'britain' in country_name or 'england' in country_name:
        return 'United Kingdom'
    elif 'canada' in country_name:
        return 'Canada'
    elif 'australia' in country_name:
        return 'Australia'
    elif 'germany' in country_name:
        return 'Germany'
    elif 'brazil' in country_name or 'brasil' in country_name:
        return 'Brazil'
    elif 'france' in country_name:
        return 'France'
    elif 'china' in country_name:
        return 'China'
    elif 'japan' in country_name:
        return 'Japan'
    elif 'russia' in country_name:
        return 'Russia'
    elif 'south africa' in country_name:
        return 'South Africa'
    elif 'mexico' in country_name:
        return 'Mexico'
    elif 'italy' in country_name:
        return 'Italy'
    elif 'spain' in country_name:
        return 'Spain'
    elif 'netherlands' in country_name or 'holland' in country_name:
        return 'Netherlands'
    elif 'new zealand' in country_name:
        return 'New Zealand'
    elif 'singapore' in country_name:
        return 'Singapore'
    elif 'sweden' in country_name:
        return 'Sweden'
    elif 'switzerland' in country_name:
        return 'Switzerland'
    else:
         return 'Other'

#Function 2:
def correct_country_name(country):
    country = str(country).strip() 
    # Try to find a close match
    match = get_close_matches(country, valid_countries, n=1, cutoff=0.8)
    if match:
        return match[0]  # Return the best match
    else:
        return  categorize_country(country)

# Apply the function to the Country column
SalaryData['Country'] = SalaryData['Country'].apply(correct_country_name)

# Verify the results
print(SalaryData['Country'].value_counts())
print(SalaryData['Country'].unique())

#show the result
display(SalaryData)
```

```python
#Final cleaning
#Drop columns that wont be used in the training model
#Working experience is kept for the charts, and left out when we split the data for training (see the train test split)
RemovedColumns=['z_score']
SalaryData=SalaryData.drop(RemovedColumns,axis=1,errors='ignore')
#Saving our work
SalaryData.to_csv(r"Datasets\SalaryDatasetFullyCleaned.csv")
#display the results 
display(SalaryData)
print("Dataset Shape:\n", SalaryData.shape)
print("\nColumn Names:\n", SalaryData.columns)
print("\nNull Counts:\n", SalaryData.isnull().sum())
print("\nData Types:\n", SalaryData.dtypes)
```

```python
# Label encoding to categorical features
from sklearn.preprocessing import LabelEncoder
features = ['Industry','Gender','Country','Education level','Job title','Race']
le = LabelEncoder()
for feature in features:
    le.fit(SalaryData[feature].unique())
    SalaryData[feature] = le.transform(SalaryData[feature])
    print(feature, SalaryData[feature].unique())
```

```python
#Train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    #We dropped the working experience because it enhance the model farther
    SalaryData.drop(['Annual salary (USD)', 'Working experince'], axis=1),
    SalaryData['Annual salary (USD)'],
    test_size=0.2,
    random_state=42
)
```

```python

from sklearn.ensemble import RandomForestRegressor

# Prepare the data for training
SalaryData['Annual salary (USD)'] = pd.to_numeric(SalaryData['Annual salary (USD)'], errors='coerce')

# Initialize the RandomForestRegressor model
rfg = RandomForestRegressor(min_samples_split=10, random_state=42)

# Fit the model
rfg.fit(X_train, y_train)

# Check the training score
train_score = rfg.score(X_train, y_train)
print(f"Training R^2 score: {train_score}")

# Check the test score
test_score = rfg.score(X_test, y_test)
print(f"Test R^2 score: {test_score}")

# Average error in dollars on the test set
from sklearn.metrics import mean_absolute_error
print(f"Test mean absolute error: {mean_absolute_error(y_test, rfg.predict(X_test)):,.0f} USD")
```
